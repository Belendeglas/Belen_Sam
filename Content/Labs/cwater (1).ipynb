````python
# Hier de data en de analyse
import numpy as np
import matplotlib.pyplot as plt
#meet een hoeveelheid water af. Voeg energie toe en meet hoe warm het water is op bepaalde tijdstippen.
#Zet deze gegevens tegen elkaar uit en bepaal zo de soortelijke weerstand. 

T_b = 20.1 + 273.15
t= np.array([0,5,10,15,20,25,30,35,40,45,50,55,55,60])
T = np.array([20.5,20.5,21.1,21.4,21.6,21.7,21.8,21.8,21.9,21.9,21.9,21.9,21.9,21.9])
T= T+ 273.15
V= 17.2
R= 1.78
m= 0.500
P = 40
plt.figure()
plt.plot(t,T)
plt.xlabel('$t$ [s]')
plt.ylabel('$T$ [K]')
plt.show()

# laatste temperatuur minus begin
deltaT = T[-1] - T[0]   # K
t_total = t[-1]         # s

c = P * t_total / (m * deltaT)

print('De gevonden soortelijke warmte van water is {:.2f} (J/kg*K)'.format(c))
   


````

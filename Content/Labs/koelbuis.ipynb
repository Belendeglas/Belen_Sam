# Theorie

Volgens [Newton's wet van afkoeling](https://nl.wikipedia.org/wiki/Afkoelingswet_van_Newton) is de snelheid waarmee een voorwerp afkoelt evenredig met het verschil in de temperatuur van het voorwerp ($T$) en de omgeving ($T_0$). We kunnen dit schrijven als:

$$
    \dot{Q} = -hA(T(t) - T_0),
$$

waarin 

- $\dot{Q}$ de warmtestroom in $\mathrm{W}$,
- $A$ het oppervlak waardoor koeling optreedt in $\mathrm{m}^2$,
- $h$ de warmteoverdrachtscoëfficiënt in $\mathrm{W/(m^2 K)}$.

Dit levert de differentiaalvergelijking

$$
    C\dot{T} = -hA(T(t) - T_0),
$$

met $C$ de warmtecapaciteit in $\mathrm{J/kg}$. Herschrijven met $\tau = \frac{C}{hA}$ levert:

$$
    -\tau\dot{T} = T(t) - T_0,
$$ (eq:dv1)

met als oplossing:

$$
    T(t) - T_0 = (T(0) - T_0)\text{e}^{-t/\tau}.
$$ (eq:dv1_opl)

We kunnen hieruit dus concluderen dat $\tau$ de karakteristieke tijdsduur is waarin de temperatuur van de buis een factor $\text{e}$ verlaagd ten opzichte van de omgevingstemperatuur. 

```{exercise} Oplossing DV controleren 
:label: ex-dvopl
Controleer bovenstaande door vergelijking [](#eq:dv1_opl) te substitueren in [](#eq:dv1).
```

$$
    T(t) = (T(0) - T_0)\text{e}^{-t/\tau} + T_0
$$
$$
    -\tau\dot{T} = (T(0) - T_0)\text{e}^{-t/\tau} + T_0 - T_0
$$
$$
    -\tau\dot{T} = (T(0) - T_0)\text{e}^{-t/\tau}
$$
 

We zijn hier voor het gemak uitgegaan van een $h$ die onafhankelijk is van de temperatuur. We weten echter dat warmtetransport door straling niet lineair gaat, maar als 

$$
    \dot{Q}_s = \epsilon \sigma A (T^4 - T_0^4).
$$ 

Voor kleine temperatuurverschillen ($\Delta T = T - T_0$) is dit te vereenvoudigen tot

$$
    \dot{Q}_s = \epsilon \sigma A ((T_0+\Delta T)^4 - T_0^4) \approx \epsilon \sigma 4A T_0^3 \Delta T.
$$

Zolang $\Delta T$ dus relatief klein is ten opzichte van $T_0$, kunnen we $h$ dus inderdaad als een constante beschouwen.

# Resultaten

````python
# Hier de data en de analyse voor geslotene buis

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def exp_func(t, A, tau, T_omg):
    # A is verschiltemperatuur met omgeving aan start
    # tau is de karakteristieke tijd voor de koeling
    # T_omg is de omgevingstemperatuur
    return (A * np.exp(-t/tau) + T_omg)  

buitenoppervlak = 3.016*10**-4 # bepaal zelf in m^2
warmtecapaciteit = 1.0749*10**2 # bepaal de warmtecapaciteit in J/K

times = np.array([5,10,15,20,25,30,35,40,45,50,60,70,80,90,100,120,130,140,150,160,170,180,200,220,240,260,280,300,320,340,360])
temps = np.array([41.6,41.4,41.1,40.9,40.7,40.2,40.2,39.9,39.3,39.5,39.3,38.2,38.1,37.8,37.1,36.7,36.5,36,35.5,35.3,35.1,34.9,34.5,34,33.5,33,32.8,32.2,31.7,31.5,31.4])


# pas beginwaardes aan naar schatting
# Het aantal maxfev moet wellicht hoger voor goede convergentie van de waarde
popt, pocv = curve_fit(exp_func, times, temps, p0=[50, 1000, 20], maxfev=5000)

A_exp, tau_exp, T_omg_exp = popt

y_fit = exp_func(times, *popt)

plt.figure()
plt.xlabel('Time [s]')
plt.ylabel('Temperature [K]')

plt.plot(times, temps, 'bo', label='measurement')
plt.plot(times, y_fit, 'r-', 
         label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp, tau_exp, T_omg_exp))
plt.title('buis with lid')

plt.legend()

plt.show()

h_exp = (warmtecapaciteit) / (tau_exp * buitenoppervlak)
 
print('the heat transfer coefficient with the lid is',h_exp) # warmteoverdrachtscoëfficiënt in W/m^2 K

times2 = np.array([0,20,40,60,80,100,157,200,230,260,290,320,350,440,470,480,510,540,570,600,630,660,690,720,750,780,810,840,870,900])
temps2 = np.array([60.5,59.4,57,54.5,52.6,50.7,46,43,41.2,30.6,38.4,37.2,36.2,35.2,34.3,32.7,32.1,31.4,31,30.4,30.2,29.7,29.4,29,28.7,28.4,28.2,28,27.8,27.7])

popt, pocv = curve_fit(exp_func, times2, temps2, p0=[50, 1000, 20], maxfev=5000)

A_exp2, tau_exp2, T_omg_exp2 = popt

y_fit2 = exp_func(times2, *popt)

plt.figure()
plt.xlabel('Time [s]')
plt.ylabel('Temperature [K]')

plt.plot(times2, temps2, 'bo', label='measurement')
plt.plot(times2, y_fit2, 'r-', 
         label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp2, tau_exp2, T_omg_exp2))
plt.title('buis without lid')

plt.legend()

plt.show()

h_exp2 = (warmtecapaciteit) / (tau_exp2 * buitenoppervlak)
 
print('the heat transfer coefficient without the lid is',h_exp2) # warmteoverdrachtscoëfficiënt in W/m^2 K
````

# Discussie en conclusie


The negative heat tranfer coefficent of the secomd experiment involving the cap indicates that rather than transfering the heat to its surroundings the metal is now recieving heat from its surroundings. the small value of it also indicates that the effiecency of the heat transfer is rather poor in nature. This points to the metal now being insulated rather than a conducting heat as normally. This lines up with the expected results as now with the cap there was far less surface area for conductivity to take place and tranfer thermal energy via this method.

The second experimental run where the insulating cap was removed is far more puzzeling as the negative value of the coefficient indicates that far more heat is being transfered to the tube than is being transfered out which is not aligning with theoretical expectations, same goes for it's small value, indicating that the metal is an insulator which it is not. possible reasons for this discrepancy cound lie in the data, the recoreded results is mared with inconsistent outliers and the time intervals are indiscriminant in their steps and take place over a far greater timeframe than with the first experiment, this cound lead to errors in the data that have yet to be accounted for.
